```python
import pandas as pd
dr = pd.read_excel('sample_-_superstore.xls')
dr = dr.drop_duplicates()
print(dr.info())
print(dr.isnull().sum())

dr['Order Date'] = pd.to_datetime(dr['Order Date'])
dr['Profit Margin'] = dr['Profit'] / dr['Sales']
dr['Order Month'] = dr['Order Date'].dt.month
dr['Order Year'] = dr['Order Date'].dt.year

topregions = dr.groupby('Region')['Sales'].sum().sort_values(ascending=False)
print(topregions)

categories = dr.groupby('Category')['Profit'].sum().sort_values(ascending=False)
print(categories)

monthlysales = dr.groupby(['Order Year', 'Order Month'])['Sales'].sum()
print(monthlysales)

monthlyprofit = dr.groupby(['Order Year', 'Order Month'])['Profit'].sum()
print(monthlyprofit)

revenue = dr.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False).head(10)
print(revenue)

profit = dr.groupby('Customer Name')['Profit'].sum().sort_values(ascending=False).head(10)
print(profit)

dr.to_excel('cleanedsuperstore.xlsx', index=False)
```
